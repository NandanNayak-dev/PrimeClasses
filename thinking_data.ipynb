```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("store_data.json")
print(data)
```

```python
def clean_data(data):
    text_to_num = {
        "one": 1,
        "two": 2,
        "three": 3,
        "four": 4,
        "five": 5
    }

    cleaned_data = []
    unique_users = set()

    for user in data:

        # Clean rating
        raw_rating = user["rating"]

        if raw_rating in text_to_num:
            user["rating"] = text_to_num[raw_rating]

        # Handle missing age
        user["age"] = user.get("age")

        # Remove duplicate users
        if user["name"] in unique_users:
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)
    print(cleaned_data)
    return cleaned_data
```

```python
data=clean_data(data)
```

```python
def get_insights(data):

    #Average Rating
    tot_rating=0
    for user in data:
        tot_rating+=float(user["rating"])
    print(f"Average rating = {tot_rating/len(data)}")

    #Percentage of poor ratings
    poor_ratings=0
    for user in data:
        if(float(user["rating"])<3):
            poor_ratings+=1
    print(f"% of poor ratings = {(poor_ratings/len(data))*100} ")
        
```

```python
get_insights(data)
```

```python
def get_recommendation(data):
    recommendations = []

    for user in data:
        curr_recomm = {}

        curr_recomm["name"] = user["name"]

        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendation(data)
```
