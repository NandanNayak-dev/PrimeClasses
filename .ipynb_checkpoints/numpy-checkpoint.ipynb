```python
import numpy as np
```

```python
arr=np.array([[1,2,5],[2,3,4]])
print((arr).shape)
```

```python
arr1=np.ones((2,3),dtype="int64")
print(arr1)
arr2=np.full((3,4),100)
print(arr2)
print(np.eye(3))
print(np.linspace(0,100,4))

print(arr.size)
```

```python
arr3=np.array([1,2,3,4])
f_arr=arr3.astype(np.float64)
print(f_arr)
```

```python
li=[1,2,3,4]
na=np.array(li)
print(na)
```

```python
lo=na.tolist()
print(type(lo))
print(type(na))
```

```python
#operations on arrays
arr=np.array([[1,2,3],[4,5,6]])
print(arr,arr.shape)
```

```python
reshaped=arr.reshape((1,6))
print(reshaped)
```

```python
flattened=arr.flatten()
print(flattened)
```

```python
#Indexing
arr=np.array([[1,2,3,4,5],[1,2,4,6,7]])
print(arr[1][4])
```

```python
arr=np.array([1,2,3,4,5])
print(arr[[1,2,3]])
```

```python
arr=np.array([1,2,3,4,5])
print(arr[arr>3])
```

```python
#Slicing
arr=np.array([1,2,3,4,5])
print(arr[1:4:2])
```
